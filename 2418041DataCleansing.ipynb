# Membaca lokasi untuk file dataset :

# Melihat Struktur Dataset

```python
print("Ukuran dataset:", data.shape)
print("\nNama kolom:")
print(data.columns.tolist())

print("\nTipe data:")
print(data.dtypes)
```

# Mengecek Data Kosong

```python
print("Jumlah data kosong:")
print(data.isnull().sum())
```

# Mengubah data kosong menjadi NaN.

```python
data = data.replace('-', pd.NA)

print("Jumlah data kosong setelah pengecekan:")
print(data.isnull().sum())
```

# Mengecek data duplikat

```python
print("Jumlah data duplikat:", data.duplicated().sum())
data[data.duplicated(keep=False)]
```

# Mengecek data yang tidak konsisten

```python
print("Nilai Jenis Kelamin:")
print(data['Jenis Kelamin'].value_counts())
print("Nilai Dusun:")
print(data['Dusun'].value_counts())
print("Nilai Status Keaktifan:")
print(data['Status Keaktifan'].value_counts())
```

# Membersihkan Nama

```python
data['Nama'] = data['Nama'].str.title()

print(data['Nama'].unique())
```

# Membersihkan Jenis Kelamin

```python
data['Jenis Kelamin'] = data['Jenis Kelamin'].replace({
    'L': 'Laki-laki',
    'Laki-laki': 'Laki-laki',
    'P': 'Perempuan',
    'Perempuan': 'Perempuan'
})

print(data['Jenis Kelamin'].value_counts())
```

# Membersihkan data Dusun

```python
data['Dusun'] = (
    data['Dusun']
    .str.replace('Dusun ', '', regex=False)
    .str.strip()
    .str.title()
)

print(data['Dusun'].value_counts())
```

# Membersihkan Status Keaktifan

```python
data['Status Keaktifan'] = data['Status Keaktifan'].str.title()

print(data['Status Keaktifan'].value_counts())
data['Status Keaktifan'] = data['Status Keaktifan'].replace({
    'Ya': 'Aktif',
    'Aktif': 'Aktif',
    'Cukup Aktif': 'Cukup Aktif',
    'Tidak Aktif': 'Tidak Aktif'
})

print(data['Status Keaktifan'].value_counts())
```

# Memperbaiki kolom Kehadiran Kegiatan

```python
def ubah_ke_persen(x):
    if pd.isna(x):
        return np.nan

    if x <= 1:
        return x * 100

    return x

data['Kehadiran Kegiatan'] = data['Kehadiran Kegiatan'].apply(ubah_ke_persen)

print(data['Kehadiran Kegiatan'])
```

# Menangani nilai yang tidak masuk akal

```python
data.loc[
    (data['Kehadiran Kegiatan'] < 0) |
    (data['Kehadiran Kegiatan'] > 100),
    'Kehadiran Kegiatan'
] = np.nan
print(data['Kehadiran Kegiatan'])
```

# Mengisi missing value

```python
median_kehadiran = data['Kehadiran Kegiatan'].median()

data['Kehadiran Kegiatan'] = data['Kehadiran Kegiatan'].fillna(
    median_kehadiran
)

print("Median kehadiran:", median_kehadiran)
print("Missing value setelah diperbaiki:")
print(data.isnull().sum())
```

# Membersihkan No HP

```python
data['No HP'] = data['No HP'].astype('string')

data['No HP'] = (
    data['No HP']
    .str.replace(r'[\s\-\.]', '', regex=True)
)

def perbaiki_no_hp(no):
    if pd.isna(no):
        return np.nan

    no = str(no)

    if no.startswith('8'):
        no = '0' + no

    return no

data['No HP'] = data['No HP'].apply(perbaiki_no_hp)

print("No HP setelah dibersihkan:")
print(data['No HP'])
```

# Menyimpan Dataset Hasil Cleansing

```python
output_path = '/content/drive/MyDrive/Data Mining/Data Hasil Cleansing Elen.xlsx'
data.to_excel(output_path, index=False)
print("Data hasil cleansing berhasil disimpan.")
```
